```python
# ==============================
# FINAL DATA QUALITY CHECK
# ==============================

# 1. Check all tables
for name, df in clean_tables.items():
    print("=" * 60)
    print(name)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Missing Values:", df.isnull().sum().sum())
    print("Duplicate Rows:", df.duplicated().sum())


# 2. Check Orders -> Customers
missing_customers = ~orders_clean["customer_id"].isin(
    customers_clean["customer_id"]
)

print("\nOrders with missing customer:", missing_customers.sum())


# 3. Check Order Items -> Orders
missing_orders = ~order_items_clean["order_id"].isin(
    orders_clean["order_id"]
)

print("Order Items with missing order:", missing_orders.sum())


# 4. Check Order Items -> Products
missing_products = ~order_items_clean["product_id"].isin(
    products_clean["product_id"]
)

print("Order Items with missing product:", missing_products.sum())


# 5. Check Order Items -> Sellers
missing_sellers = ~order_items_clean["seller_id"].isin(
    sellers_clean["seller_id"]
)

print("Order Items with missing seller:", missing_sellers.sum())


# 6. Check Payments -> Orders
missing_payment_orders = ~order_payments_clean["order_id"].isin(
    orders_clean["order_id"]
)

print("Payments with missing order:", missing_payment_orders.sum())


# 7. Check Reviews -> Orders
missing_review_orders = ~order_reviews_clean["order_id"].isin(
    orders_clean["order_id"]
)

print("Reviews with missing order:", missing_review_orders.sum())
```

```python
# ==============================
# EXPORT CLEANED DATA
# ==============================

for name, df in clean_tables.items():
    df.to_csv(f"cleaned_{name}.csv", index=False)

print("All cleaned files exported successfully!")
```
